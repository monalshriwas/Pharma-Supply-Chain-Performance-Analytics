### Product Demand

```python
product_demand = (
    df.groupby("product_name")["quantity"]
    .sum()
    .sort_values(ascending=False)
)

print(product_demand.head(10))
```

### Revenue Analysis

```python
# Created new column 'revenue'
df["revenue"] = df["quantity"] * df["unit_price"]
```

```python
revenue_by_product = (
    df.groupby("product_name")["revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(revenue_by_product)
```

### Product summary

```python
product_performance_summary= (
    df.groupby("product_name")
      .agg(
          total_revenue=("revenue", "sum"),
          delay_rate=("delay_flag", "mean"),
          waste_rate=("waste_flag", "mean"),
          temp_risk_rate=("temp_excursion_flag", "mean"),
          total_quantity=("quantity", "sum")
      )
)

product_performance_summary["delay_rate"] *= 100
product_performance_summary["waste_rate"] *= 100
product_performance_summary["temp_risk_rate"] *= 100
product_performance_summary = product_performance_summary.round(2)

print(product_performance_summary.sort_values("total_revenue", ascending=False))
```

### Waste 

```python
df["waste_cost"] = df["waste_flag"] * df["revenue"]
waste_by_product = (
    df.groupby("product_name")["waste_cost"]
      .sum()
      .sort_values(ascending=False)
)

print(waste_by_product)
```

### Zone Wise Delay
